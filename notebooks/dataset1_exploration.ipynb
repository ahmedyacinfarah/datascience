```python
!kaggle datasets download -d fedesoriano/heart-failure-prediction -q
!unzip -q heart-failure-prediction.zip -d dataset1

df1 = pd.read_csv('dataset1/heart.csv')
df1.head()
```

```python
### Dataset 1: Heart Failure Prediction
- **Author**: Fedesoriano
- **Purpose**: Predict heart failure risk based on patient attributes
- **Shape**: 918 rows × 12 columns
- **Features**:
  - age (numerical)
  - anaemia (categorical: 0/1)
  - creatinine_phosphokinase (numerical)
  - diabetes (categorical: 0/1)
  - ejection_fraction (numerical)
  - high_blood_pressure (categorical: 0/1)
  - platelets (numerical)
  - serum_creatinine (numerical)
  - serum_sodium (numerical)
  - sex (categorical: 0=Female, 1=Male)
  - smoking (categorical: 0/1)
  - DEATH_EVENT (categorical: 0=No, 1=Yes)
- **Missing values**: None
```

```python
# r1: Distribution of Age (Numerical Univariate)
plt.figure(figsize=(8,5))
sns.histplot(df1['Age'], bins=20, kde=True)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Count")
plt.show()
```

```python
**Insight 1:** Most patients are between 50 and 70 years old.
**Analysis:** Univariate numerical distribution (r1)
**Code:** Shown above.
```

```python
### Dataset 2: Student Performance
- **Author**: Ayesha Siddiqa
- **Purpose**: Understand factors affecting student exam scores
- **Shape**: [rows] × 20 columns
- **Features** (subset):
  - Hours_Studied (numerical)
  - Attendance (numerical)
  - Parental_Involvement (numerical)
  - Access_to_Resources (numerical)
  - Motivation_Level (numerical)
  - Gender (categorical)
  - Exam_Score (numerical)
- **Missing values**: [if any]
```
